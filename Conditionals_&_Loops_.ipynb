```python
# Declare the number
number = 0

# Check the value of the number
if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```

```python
# Declare the person's age
age = 45

# Determine the age group using conditional statements
if age < 13:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 59:
    print("Adult")
else:
    print("Senior")
```

```python
# Declare the character
character = "e"

# Convert to lowercase to handle both uppercase and lowercase inputs easily
char_lower = character.lower()

# Check if the character is in the list of vowels
if char_lower in ['a', 'e', 'i', 'o', 'u']:
    print("Vowel")
else:
    print("Consonant")
```

```python
# Declare the number
number = 55

# Check if the number is divisible by both 5 and 11
if number % 5 == 0 and number % 11 == 0:
    print("Divisible by 5 and 11")
else:
    print("Not divisible by 5 and 11")
```

```python
# Declare n
n = 5

# Initialize a variable to store the sum
total_sum = 0

# Loop from 1 to n (inclusive)
for i in range(1, n + 1):
    total_sum += i

# Print the result matching the sample output
print(f"Sum = {total_sum}")
```

```python
# Declare the number
number = 6

# Loop from 1 to 10 to generate the multiplication table
for i in range(1, 11):
    product = number * i
    print(f"{number} x {i} = {product}")
```

```python
# Declare the number
number = 5

# Initialize the factorial variable to 1
factorial = 1

# Loop from 1 to the given number (inclusive)
for i in range(1, number + 1):
    factorial *= i

# Print the result matching the sample output
print(f"Factorial = {factorial}")
```

```python
# Declare the number
number = 1234

# Initialize variables
reversed_num = 0
temp = number

# Loop to extract and reverse digits
while temp > 0:
    digit = temp % 10  # Get the last digit
    reversed_num = (reversed_num * 10) + digit  # Append digit to the reversed number
    temp = temp // 10  # Remove the last digit from temp

# Print the result matching the sample output
print(f"Reversed Number = {reversed_num}")
```

```python
# Declare the number of terms
n = 10

# Initialize the first two terms of the Fibonacci series
a, b = 0, 1

# Print the Fibonacci series
print("Fibonacci Series:")
for i in range(n):
    print(a, end=" " if i < n - 1 else "")
    # Update values for the next term
    next_term = a + b
    a = b
    b = next_term
```

```python
# Declare the number
number = 12345

# Initialize count and a temporary variable
count = 0
temp = number

# Handle the edge case if the number is 0
if number == 0:
    count = 1
else:
    # Loop to remove digits one by one until temp becomes 0
    while temp > 0:
        count += 1
        temp = temp // 10

# Print the result matching the sample output
print(f"Number of digits = {count}")
```
