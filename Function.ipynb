```python
def is_even(n):
    return n % 2 == 0

print(is_even(10))  # should print True
print(is_even(7))   # should print False

# Stretch goal: write is_odd(n) that uses is_even[cite: 11]
def is_odd(n):
    return not is_even(n)

print(is_odd(10))   # should print False
print(is_odd(7))    # should print True
```

```python
def greet(name, language="English"):
    if language == "Hindi":
        return "Namaste, " + name
    elif language == "Tamil":
        return "Vanakkam, " + name
    else:
        return "Hello, " + name

print(greet("Sathish"))
print(greet("Sathish", "Hindi"))
print(greet("Sathish", language="Tamil"))
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        return self.balance

acc1 = BankAccount("Sathish", 1000)
acc2 = BankAccount("Kumar", 2500)

print(acc1.owner, "Deposit:", acc1.deposit(500))
print(acc2.owner, "Deposit:", acc2.deposit(1000))
```

```python
def snack_discount(total_tickets):
    if total_tickets >= 4:
        return 50
    else:
        return 0

def book_tickets(ticket_price, count, has_coupon=False):
    subtotal = ticket_price * count
    if has_coupon:
        subtotal = subtotal * 0.85

    discount = snack_discount(count)
    final_amount = subtotal - discount
    return round(final_amount, 2)

print(book_tickets(200, 5, has_coupon=True))
print(book_tickets(200, 2, has_coupon=False))
```
