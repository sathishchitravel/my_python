```python
# Sample input - already filled in for you.
target = 40
runs_per_over = [9, 6, 12, 4, 11, 7]
score = 0
overs_used = 0

for runs in runs_per_over:
    # TODO: add runs to score and add 1 to overs_used
    score += runs
    overs_used += 1

    # TODO: print the trace line for this over
    print("Over", overs_used, "score", score)

    # TODO: stop the loop once score reaches target
    if score >= target:
        break

# Result check - already filled in for you.
if score >= target:
    print("Target reached in", overs_used, "overs")
else:
    print("Target not reached. Final score:", score)
```

```python
BASE_FARE = 30.0
RIDES = [(1.5, False), (5, False), (5, True), (8.4, True), (10, False)]


def extra_km_charge(km, rate_per_km=15):
  if km <= 2:
    return 0
  return (km - 2) * rate_per_km


def auto_fare(km, night=False, rate_per_km=15):
  fare = BASE_FARE + extra_km_charge(km, rate_per_km=rate_per_km)
  if night:
    fare *= 1.5
  return round(fare, 2)


for km, night in RIDES:
  print(f"{km} km, night={night}: {auto_fare(km, night=night)}")
```
