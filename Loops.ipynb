```python
start = 1
end = 10
for i in range(start, end + 1):
    print(i)
```

```python
n = 5
total_sum = 0
for i in range(1, n + 1):
    total_sum += i
print(f"Sum = {total_sum}")
```

```python
number = 7
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

```python
limit = 20
stop_at = 13
for i in range(1, limit + 1):
    if i == stop_at:
        break
    print(i)
```

```python
limit = 20
skip = 3
for i in range(1, limit + 1):
    if i % skip == 0:
        continue
    print(i)
```

```python
rows = 5
for i in range(1, rows + 1):
    print("*" * i)
```

```python
first = 1
last = 3
upto = 10
for table in range(first, last + 1):
    print(f"Table of {table}")
    for i in range(1, upto + 1):
        print(f"{table} x {i} = {table * i}")
```

```python
secret = 7
guesses = [3, -1, 5, 7]
for guess in guesses:
    if guess < 0:
        print("Negative numbers are not allowed. Try again.")
        continue
    elif guess == secret:
        print("Congratulations! You guessed it correctly.")
        break
    else:
        print("Wrong guess. Try again.")
```
